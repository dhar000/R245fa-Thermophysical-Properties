```python
import CoolProp.CoolProp as CP
import numpy as np
import pandas as pd

# Fluid and Supercritical Pressure (3.651 MPa)
fluid = "R245fa"
P_Pa = 36.51 * 1e5

# Temperature range from 70°C to 160°C
T_C = np.linspace(70, 160, 91)
T_K = T_C + 273.15

density, cp_list, conductivity, viscosity = [], [], [], []

for t in T_K:
  density.append(CP.PropsSI("D", "T", t, "P", P_Pa, fluid))
  cp_list.append(CP.PropsSI("Cpmass", "T", t, "P", P_Pa, fluid))
  conductivity.append(CP.PropsSI("conductivity", "T", t, "P", P_Pa, fluid))
  viscosity.append(CP.PropsSI("viscosity", "T", t, "P", P_Pa, fluid))

df = pd.DataFrame({
    "Temperature (°C)": T_C,
    "Temperature (K)": T_K,
    "Density (kg/m³)": density,
    "Specific Heat, Cp (J/kg·K)": cp_list,
    "Thermal Conductivity (W/m·K)": conductivity,
    "Viscosity (Pa·s)": viscosity,
})

df.to_excel("R245fa_Supercritical_Properties.xlsx", index=False)
print("Data saved successfully to R245fa_Supercritical_Properties.xlsx")
```
